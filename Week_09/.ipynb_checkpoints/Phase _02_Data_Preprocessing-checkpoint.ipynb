```python
import os
import cv2
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
```

```python
#Load Metadata
metadata = pd.read_csv("Datasets/HAM10000_metadata.csv")
```

```python
#Create Image Path
import glob

image_paths = {}

for folder in [
    "Datasets/HAM10000_images_part_1",
    "Datasets/HAM10000_images_part_2"
]:
    for path in glob.glob(os.path.join(folder, "*.jpg")):
        image_id = os.path.splitext(os.path.basename(path))[0]
        image_paths[image_id] = path

metadata["path"] = metadata["image_id"].map(image_paths)
```

```python
#Encode Target Labels
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

metadata["label"] = encoder.fit_transform(metadata["dx"])

metadata[["dx","label"]].head()
```

```python
#Train / Validation / Test Split
train_df, temp_df = train_test_split(
    metadata,
    test_size=0.30,
    stratify=metadata["label"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=42
)
```

```python
#Create Image Loader
IMG_SIZE = 224

def load_image(path, label):

    image = tf.io.read_file(path)

    image = tf.image.decode_jpeg(image, channels=3)

    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))

    image = tf.cast(image, tf.float32)

    image = image / 255.0

    return image, label
```

```python
#Create TensorFlow Datasets
train_ds = tf.data.Dataset.from_tensor_slices(
    (
        train_df["path"].values,
        train_df["label"].values
    )
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (
        val_df["path"].values,
        val_df["label"].values
    )
)

test_ds = tf.data.Dataset.from_tensor_slices(
    (
        test_df["path"].values,
        test_df["label"].values
    )
)
```

```python
#Map Image Loader
train_ds = train_ds.map(load_image)

val_ds = val_ds.map(load_image)

test_ds = test_ds.map(load_image)
```

```python
#Data Augmentation
data_augmentation = tf.keras.Sequential([

    tf.keras.layers.RandomFlip(
        "horizontal_and_vertical"
    ),

    tf.keras.layers.RandomRotation(
        0.2
    ),

    tf.keras.layers.RandomZoom(
        0.2
    ),

    tf.keras.layers.RandomContrast(
        0.2
    ),

    tf.keras.layers.RandomBrightness(
        factor=0.2
    ),

    tf.keras.layers.RandomCrop(
        200,
        200
    ),

    tf.keras.layers.Resizing(
        224,
        224
    )

])
```

```python
#Apply Augmentation Only to Training Data
train_ds = train_ds.map(
    lambda x, y: (data_augmentation(x, training=True), y)
)
```

```python
#Handle Class Imbalance
train_df["label"].value_counts()
```

```python
#Compute Class Weights
class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_df["label"]),
    y=train_df["label"]
)

class_weights = dict(enumerate(class_weights))

print(class_weights)
```

```python
#Batch and Prefetch
BATCH_SIZE = 32

AUTOTUNE = tf.data.AUTOTUNE

train_ds = (
    train_ds
    .shuffle(1000)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

val_ds = (
    val_ds
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

test_ds = (
    test_ds
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)
```

```python
#Visualize Augmented Images
plt.figure(figsize=(12, 12))

for images, labels in train_ds.take(1):

    for i in range(9):

        plt.subplot(3, 3, i + 1)

        plt.imshow(images[i])

        plt.title(f"Label: {labels[i].numpy()}")

        plt.axis("off")

plt.tight_layout()

plt.show()
```
