```python
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    BatchNormalization,
    Dropout,
    Dense,
    Flatten,
    GlobalAveragePooling2D,
    Input
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.applications import (
    MobileNetV2,
    EfficientNetB0,
    ResNet50,
    DenseNet121
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
#Load Dataset

train_df = pd.read_csv("Datasets/train.csv")
val_df = pd.read_csv("Datasets/val.csv")
test_df = pd.read_csv("Datasets/test.csv")

print(train_df.shape)
print(val_df.shape)
print(test_df.shape)
```

```python
#Encode Labels
encoder = LabelEncoder()

train_df["label"] = encoder.fit_transform(train_df["dx"])
val_df["label"] = encoder.transform(val_df["dx"])
test_df["label"] = encoder.transform(test_df["dx"])

print(encoder.classes_)
```

```python
#Convert Labels to String

train_df["label"] = train_df["label"].astype(str)
val_df["label"] = val_df["label"].astype(str)
test_df["label"] = test_df["label"].astype(str)
```

```python
#Constants
IMG_SIZE = 128
BATCH_SIZE = 16
EPOCHS = 5
NUM_CLASSES = 7
```

```python
#Class_Weights
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.unique(train_df["label"])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_df["label"]
)

class_weights = dict(zip(classes, weights))

print(class_weights)
```

```python
#Data Augmentation (same for every model)
train_datagen = ImageDataGenerator(

    rescale=1./255,

    rotation_range=20,

    horizontal_flip=True,

    vertical_flip=True,

    zoom_range=0.2,

    brightness_range=[0.8,1.2]

)

test_datagen = ImageDataGenerator(

    rescale=1./255

)
```

```python
#Train Generator
train_generator = train_datagen.flow_from_dataframe(

    dataframe=train_df,

    x_col="path",

    y_col="label",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=True

)
```

```python
#Validation Generator
val_generator = test_datagen.flow_from_dataframe(

    dataframe=val_df,

    x_col="path",

    y_col="label",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=False

)
```

```python
#Test Generator
test_generator = test_datagen.flow_from_dataframe(

    dataframe=test_df,

    x_col="path",

    y_col="label",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=False

)
```

```python
results = {
    "Model": [],
    "Train Accuracy": [],
    "Validation Accuracy": [],
    "Test Accuracy": [],
    "Precision": [],
    "Recall": [],
    "F1 Score": []
}
```

```python
# Basic CNN
basic_cnn = Sequential([

    Input(shape=(IMG_SIZE, IMG_SIZE, 3)),

    Conv2D(32, (3,3), activation="relu"),
    MaxPooling2D(2,2),

    Conv2D(64, (3,3), activation="relu"),
    MaxPooling2D(2,2),

    GlobalAveragePooling2D(),

    Dense(128, activation="relu"),

    Dense(NUM_CLASSES, activation="softmax")

])

basic_cnn.summary()
```

```python
#Compile Model
basic_cnn.compile(

    optimizer=Adam(learning_rate=0.0001),

    loss="categorical_crossentropy",

    metrics=["accuracy"]

)
```

```python
#Train Model
history_basic = basic_cnn.fit(

    train_generator,

    validation_data=val_generator,

    epochs=EPOCHS,

    class_weight=class_weights

)
```

```python
#Evaluate Model
test_loss, test_acc = basic_cnn.evaluate(test_generator)

print(f"Test Accuracy : {test_acc:.4f}")

print(f"Test Loss : {test_loss:.4f}")
```

```python
#Calculate Metrics
pred = basic_cnn.predict(test_generator)

pred_labels = np.argmax(pred, axis=1)

true_labels = test_generator.classes
```

```python
#Accuracy, Precision, Recall, F1
accuracy = accuracy_score(true_labels, pred_labels)

precision = precision_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
```

```python
train_acc =history_basic.history["accuracy"][-1]

val_acc = history_basic.history["val_accuracy"][-1]

train_loss = history_basic.history["loss"][-1]

val_loss = history_basic.history["val_loss"][-1]
print(val_acc)
```

```python
#Save Results
results["Model"].append("Basic CNN")

results["Train Accuracy"].append(history_basic.history["accuracy"][-1])

results["Validation Accuracy"].append(history_basic.history["val_accuracy"][-1])

results["Test Accuracy"].append(accuracy)

results["Precision"].append(precision)

results["Recall"].append(recall)

results["F1 Score"].append(f1)
```

```python
#Save Model
basic_cnn.save("models/basic_cnn1.keras")
```

```python
#CNN + Dropout
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    GlobalAveragePooling2D,
    Dense,
    Dropout
)

cnn_dropout = Sequential([

    Input(shape=(IMG_SIZE, IMG_SIZE, 3)),

    Conv2D(32, (3,3), activation="relu"),
    MaxPooling2D(2,2),

    Conv2D(64, (3,3), activation="relu"),
    MaxPooling2D(2,2),

    Conv2D(128, (3,3), activation="relu"),
    MaxPooling2D(2,2),

    Conv2D(256, (3,3), activation="relu"),
    MaxPooling2D(2,2),

    GlobalAveragePooling2D(),

    Dense(512, activation="relu"),
    Dropout(0.5),

    Dense(256, activation="relu"),
    Dropout(0.4),

    Dense(128, activation="relu"),
    Dropout(0.3),

    Dense(NUM_CLASSES, activation="softmax")

])

cnn_dropout.summary()
```

```python
#Compile
cnn_dropout.compile(

    optimizer=Adam(learning_rate=0.0001),

    loss="categorical_crossentropy",

    metrics=["accuracy"]

)
```

```python
#Train
history_dropout = cnn_dropout.fit(

    train_generator,

    validation_data=val_generator,

    epochs=EPOCHS,

    class_weight=class_weights

)
```

```python
#Evaluate
test_loss, test_acc = cnn_dropout.evaluate(test_generator)

print("Test Accuracy :", round(test_acc,4))

print("Test Loss :", round(test_loss,4))
```

```python
#Predict
pred = cnn_dropout.predict(test_generator)

pred_labels = np.argmax(pred, axis=1)

true_labels = test_generator.classes
```

```python
#Metrics
accuracy = accuracy_score(
    true_labels,
    pred_labels
)

precision = precision_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
```

```python
#results["Model"].append("CNN + Dropout")

results["Train Accuracy"].append(
    history_dropout.history["accuracy"][-1]
)

results["Validation Accuracy"].append(
    history_dropout.history["val_accuracy"][-1]
)

results["Test Accuracy"].append(accuracy)

results["Precision"].append(precision)

results["Recall"].append(recall)

results["F1 Score"].append(f1)
```

```python
#Build MobileNetV2
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, GlobalAveragePooling2D, Dense

base_model = MobileNetV2(

    weights="imagenet",

    include_top=False,

    input_shape=(IMG_SIZE, IMG_SIZE, 3)

)

base_model.trainable = False

inputs = Input(shape=(IMG_SIZE, IMG_SIZE, 3))

x = base_model(inputs, training=False)

x = GlobalAveragePooling2D()(x)

x = Dense(256, activation="relu")(x)

outputs = Dense(NUM_CLASSES, activation="softmax")(x)

mobilenet_model = Model(inputs, outputs)

mobilenet_model.summary()
```

```python
#Compile
mobilenet_model.compile(

    optimizer=Adam(learning_rate=0.0001),

    loss="categorical_crossentropy",

    metrics=["accuracy"]

)
```

```python
#Train
history_mobile = mobilenet_model.fit(

    train_generator,

    validation_data=val_generator,

    epochs=EPOCHS,

    class_weight=class_weights

)
```

```python
#Evaluate
test_loss, test_acc = mobilenet_model.evaluate(test_generator)

print("Test Accuracy :", round(test_acc,4))

print("Test Loss :", round(test_loss,4))
```

```python
#predictions
pred = mobilenet_model.predict(test_generator)

pred_labels = np.argmax(pred, axis=1)

true_labels = test_generator.classes
```

```python
#Metrics
accuracy = accuracy_score(
    true_labels,
    pred_labels
)

precision = precision_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    true_labels,
    pred_labels,
    average="weighted",
    zero_division=0
)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
```

```python
#Save Results
results["Model"].append("MobileNetV2")

results["Train Accuracy"].append(
    history_mobile.history["accuracy"][-1]
)

results["Validation Accuracy"].append(
    history_mobile.history["val_accuracy"][-1]
)

results["Test Accuracy"].append(accuracy)

results["Precision"].append(precision)

results["Recall"].append(recall)

results["F1 Score"].append(f1)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
import numpy as np

def evaluate_model(model,
                   model_name,
                   history,
                   test_generator,
                   results):

    # Train Accuracy
    train_acc = max(history.history["accuracy"])

    # Validation Accuracy
    val_acc = max(history.history["val_accuracy"])

    # Test Accuracy
    test_loss, test_acc = model.evaluate(test_generator, verbose=0)

    # Predictions
    y_pred = model.predict(test_generator, verbose=0)
    y_pred = np.argmax(y_pred, axis=1)

    # True Labels
    y_true = test_generator.classes

    # Metrics
    precision = precision_score(
        y_true,
        y_pred,
        average="weighted"
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="weighted"
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted"
    )

    # Store Results
    results["Model"].append(model_name)
    results["Train Accuracy"].append(train_acc)
    results["Validation Accuracy"].append(val_acc)
    results["Test Accuracy"].append(test_acc)
    results["Precision"].append(precision)
    results["Recall"].append(recall)
    results["F1 Score"].append(f1)

    print(f"{model_name} Evaluation Completed")
```

```python
evaluate_model(
    basic_cnn,
    "Basic CNN",
    history_basic,
    test_generator,
    results
)

evaluate_model(
    deep_cnn,
    "Deep CNN",
    history_deep,
    test_generator,
    results
)

evaluate_model(
    cnn_bn,
    "CNN + Batch Normalization",
    history_bn,
    test_generator,
    results
)

evaluate_model(
    cnn_dropout,
    "CNN + Dropout",
    history_dropout,
    test_generator,
    results
)

evaluate_model(
    mobilenet_model,
    "MobileNetV2",
    history_mobile,
    test_generator,
    results
)

evaluate_model(
    efficientnet_model,
    "EfficientNetB0",
    history_efficient,
    test_generator,
    results
)

evaluate_model(
    resnet_model,
    "ResNet50",
    history_resnet,
    test_generator,
    results
)

evaluate_model(
    densenet_model,
    "DenseNet121",
    history_dense,
    test_generator,
    results
)
```
