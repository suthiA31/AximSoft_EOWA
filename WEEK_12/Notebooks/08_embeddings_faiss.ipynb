```python
from sentence_transformers import SentenceTransformer
import faiss
```

```python
tickets = pd.read_csv(
    "processed/processed_support_tickets.csv"
)

print("Shape:", tickets.shape)

tickets[
    ["Ticket ID", "Ticket Type", "Ticket Subject", "CleanText"]
].head()
```

```python
tickets["SearchText"] = (
    tickets["Ticket Subject"].fillna("").astype(str)
    + ". "
    + tickets["CleanText"].fillna("").astype(str)
)

tickets["SearchText"].head()
```

```python
model_name = "all-MiniLM-L6-v2"

embedding_model = SentenceTransformer(
    model_name
)

print("Embedding model loaded.")
```

```python
start_time = time.time()

embeddings = embedding_model.encode(
    tickets["SearchText"].tolist(),
    show_progress_bar=True,
    convert_to_numpy=True
)

embedding_time = time.time() - start_time

print("Embedding shape:", embeddings.shape)
print(f"Embedding time: {embedding_time:.2f} seconds")
```

```python
faiss.normalize_L2(embeddings)

print("Embeddings normalized.")
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(
    dimension
)

index.add(
    embeddings
)

print("FAISS index created.")
print("Number of vectors:", index.ntotal)
```

```python
query = "My product is not working properly"

query_embedding = embedding_model.encode(
    [query],
    convert_to_numpy=True
)

faiss.normalize_L2(query_embedding)
```

```python
distances, indices = index.search(
    query_embedding,
    5
)
```

```python
results = tickets.iloc[
    indices[0]
][[
    "Ticket ID",
    "Ticket Type",
    "Ticket Subject",
    "Ticket Description"
]].copy()

results["Similarity"] = distances[0]

results
```

```python
def semantic_search(query, top_k=5):

    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    )

    faiss.normalize_L2(query_embedding)

    distances, indices = index.search(
        query_embedding,
        top_k
    )

    results = tickets.iloc[
        indices[0]
    ].copy()

    results["Similarity"] = distances[0]

    return results
```

```python
semantic_search(
    "I have a problem with my bill",
    top_k=5
)
```

```python
semantic_search(
    "I cannot access my account",
    top_k=5
)
```

```python
semantic_search(
    "My computer has a technical problem",
    top_k=5
)
```
